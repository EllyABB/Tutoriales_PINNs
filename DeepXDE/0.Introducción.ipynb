## Como calcular derivadas:

Differenciación automática. Autodiff

1. Jacobian
    - Para una función escalar f(x) el Jacobiano es simplemente la primera derivada 
    - para una funcion multivariable f(x,y), el jacobiano es un vector de las derivadas parciales respecto a cada variable
2. Hessian
    - El Hesiano es una matris de segundas derivadas parciales
    - Para una función escalar f(x) es solo la segunda derivada respecto a x
    - Para una función f(x,y) es una matris que incluye todas las posibles segundas derivadas

<br> 

### Por ejemplo: <br> 
Para una función $f(x,y)=x^2+xy+y^2$,<br> 

1. Jacobiano:<br> 
    - .  $\frac{\partial f}{\partial x} = 2x+y$<br> 
    - . $\frac{\partial f}{\partial y} = x+2y$<br> 
    El vector sería $$Jacobian = \left[ \begin{matrix}
2x + y\\
x + 2y 
\end{matrix}\right]$$

2. Hessiano:<br> 
    - .  $\frac{\partial^2 f}{\partial x^2} = 2$<br> 
    - . $\frac{\partial^2 f}{\partial y^2} = 2$<br>  
    - . $\frac{\partial^2 f}{\partial x \partial y} = 1$<br> 
    - . $\frac{\partial^2 f}{\partial y \partial x} = 1$<br> 
    El vector sería $$Hessian = \left[ \begin{matrix}
2 & 1\\
1 & 2 
\end{matrix}\right]$$


> dde.grad.hesian(y,x)

**3. RobinBC**

Las condiciones de frontera Robin son una mexcla de las condiciones de frontera de dirichlet y las de neumann.
Estas condiciones de frontera  (Junto con la BC periódica) son tareas más críticas que aveces incluso pueden introducir inestabilidades en el entrenamiento lo que puede llebar a una solucion de la red debil.

```Python
dde.icbc.RobinBC(geom, boundary_func,boundary)
```
Donde:
    
```Python
def boundary_flux_r(x,y):  #<--- Ahora tiene dos argumentos
    return y #Sets du/dx=u at all x=1

def boundary_r(x,on_boundary):
    return on_boundary and dde.utils.isclose(x[0],1)
```
La función de frontera, **boundary_func**
Acá se fija $$a\cdot u + b\cdot \frac{du}{dx}=g$$ donde $a$, $b$, $g$ son constantes o funciones.



**4. PeriodicBC**

Las condiciones de frontera periodicas implican que la solucion debe tener el mismo valor en puntos pares atravez de fronteras opuestas, creando una solucion cíclica.

Estas condiciones de frontera  (Junto con la BC Robin) son tareas más críticas que aveces incluso pueden introducir inestabilidades en el entrenamiento lo que puede llebar a una solucion de la red debil.

```Python
dde.icbc.PeriodicBC(geom, component_x, on_boundary,derivative_order=0,component=0)
```
Donde:
```Python
component_x  # <---- Especifica el eje o la dimension espacial a lo largo de la cual se pide la periodicidad
             #       No es una coordenada específica
             #       En una geometria tipo rectangulo (x,y), component_x=0 pone periodicidad en el eje x
             #                                             y component_x=1 pone periodicidad en el eje y
on_boundary  # <---- La misma función boleana de antes.

derivative_order=0 # <---- Por defaut = 0. Especifica si la condición de periodicidad es sobre la función 
                   #       o sobre algun orden de derivada. Si = 1 => u'(a)=u'(b) 
component=0  # <---- Si se soluciona una PDE con más incognitas, ej. u,v. component=0 -> u(x,y)
             #                                                            component=1 -> v(x,y)
``` 


##### Ejemplo:

```Python

    return on_boundary

```

Si queremos condiciones periodicas sobre puntos que no son la frontera, por ejemplo, $x\in[-1,1]$ y $u(-1)=u(0)$, ya no se puede usar este objeto. Se usan otras BC llamadas OperatorBC.





**5. OperatorBC**


Son tambien conocidas como condiciones de frontera no estandar.<br>
Algunos ejemplos serian:
1. Higher - Order Boundary Conditions
$$u''(1)=0,\qquad u'''(1)=0$$
2. Nonlinear Boundary Conditions
$$u'(1)+u(1)^2=0,$$
3. Mixed-Type Conditions
$$u'(0)+\sin(u(1))=0,$$

Este tipo de condiciones no puede ser usado por ninguna forma estandar.


La clase OperatorBC permite forzar condiciones de frontera generales de un cierto operador. A diferencia de las condiciones de frontera estandar, este deha definir constricciones personalizadas que involucran operaciones en las entradas salidas y derivadas. Es de la forma $$f(inputs,outputs,X)=0$$ donde:
- **inputs**: Valores de entrada del modelo (ej. Coordenadas espaciales)
- **outputs**: Valores de predichos por el modelo 
- **x**: El arreglo de los puntos inputs originales

**Los inputs/outputs deben darse en la forma de un tensor dependiendo de la libreria. (Pytorch o Tensorflow)**

#### Syntax

```Python
dde.icbc.OperatorBC(geom, func, on_boundary)
```
Donde:
- **func** Es la funcion que define la condicion de frontera. Esta debe tomar 3 argumentos. Inputs, outputs y x. La función debe retornat una tensor del tipo $N \times 1$ donde N es el número de puntos en la frontera.
- on_boundary. Una funcion boleana como las anteriores con x el punto y on_boundary el valor boleano.

## Number of Collocation Points (Points where the network is trained to satisfy the PDE	)

Training


El valor 16 se refiere al número de puntos de colocación por batch. Estos son los puntos interiores muestreados dentro del dominio definido por geom, donde la red neuronal se entrena para satisfacer la ecuación diferencial. Los puntos de colocación se utilizan para calcular los residuos de la PDE durante el entrenamiento; es decir, el grado en que la solución de la red neuronal satisface la PDe en esos puntos específicos. La red minimiza estos residuos mediante retropropagación.


Por lo tanto, los puntos de colocación no evalúan la precisión; son los "datos de entrenamiento" generados a partir de la EDP y la geometría.

## num_test=100 (Points where the model is evaluated against the exact solution	)


Evaluation / Testing


Especifica el número de puntos de prueba, es decir, los puntos utilizados para evaluar el rendimiento del modelo entrenado. En estos 100 puntos:

El modelo entrenado predice valores. Estas predicciones se comparan con los valores reales (de solution=func, la solución exacta o conocida). Se calculan métricas como el error L2 para evaluar la precisión del modelo.
